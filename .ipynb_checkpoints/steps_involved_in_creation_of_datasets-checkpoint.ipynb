# Mufulira Municipal Council Data Extraction

## Overview

This project involved collecting and transforming publicly available information from the **Mufulira Municipal Council** website into structured datasets for data mining and analysis.

Two major data categories were extracted:

1. **Constituency Development Fund (CDF) Community Projects**
2. **Budgets and Revenue Streams**

The extraction process was designed to preserve the original information from the council's documents while converting it into structured, machine-readable CSV datasets.

The final datasets use the pipe character (`|`) as the delimiter rather than the standard comma delimiter.

---

# 1. Constituency Development Fund (CDF)

## 1.1 CDF Data Source

CDF community project information was obtained from the Mufulira Municipal Council's official CDF page:

```text
https://www.mufuliracouncil.gov.zm/?page_id=792

The page contains links to Community Projects PDF documents for different Mufulira constituencies.

The extraction focused on the following constituencies:

Kantanshi
Kankoyo
Mufulira Central

Rejected project documents were excluded from the extraction.

1.2 Discovering CDF PDF Documents

The CDF scraper does not require the PDF files to already exist in the local project directory.

Instead, the Python script first accesses the official Mufulira Municipal Council CDF webpage and searches for PDF links.

The scraper uses:

requests to retrieve the webpage
BeautifulSoup to parse the webpage HTML
urljoin to construct complete PDF URLs
Regular expressions and string matching to identify the relevant documents

The scraper checks that each discovered link:

Points to a PDF file.
Contains comm-projects in the URL.
Is not a rejected project document.
Can be associated with a known constituency.

The constituency is determined from the PDF filename.

For example:

Kantanshi-Comm-Projects.pdf
Kankoyo-Comm-Projects.pdf
Mufulira-Central-Comm-Projects.pdf

This allows the scraper to automatically identify the appropriate constituency for each document.

1.3 Downloading CDF PDFs Into Memory

The CDF PDFs are not saved to the local project directory.

Instead, the scraper downloads each PDF directly from the council website and keeps the PDF content in memory.

The PDF is obtained using:

response = session.get(
    url,
    timeout=60,
    verify=False
)

response.raise_for_status()

pdf_bytes = response.content

The PDF is therefore available as a byte object and can be processed immediately.

The extraction process is:

Mufulira Municipal Council CDF Page
                |
                v
        Discover PDF Links
                |
                v
      Identify Constituency
                |
                v
      Download PDF Into Memory
                |
                v
          OCR Processing
                |
                v
       Structured CDF Records
                |
                v
        Pipe-Delimited CSV

No CDF PDF files are permanently stored in the project directory.

1.4 OCR Processing

The CDF Community Projects PDFs are scanned documents.

Because the project information is contained primarily in scanned images, normal text extraction is not sufficient.

Optical Character Recognition (OCR) is therefore used.

The Python script uses:

pdf2image
pytesseract
Tesseract OCR

The PDF bytes are converted into images at 300 DPI:

images = convert_from_bytes(
    pdf_bytes,
    dpi=300
)

Each page is then passed to Tesseract:

ocr_text = pytesseract.image_to_string(img)

This converts the scanned project information into text that can be processed programmatically.

1.5 Page Orientation Handling

Some scanned pages may be incorrectly rotated.

The scraper attempts to detect the page orientation using Tesseract's orientation detection.

If a rotation angle is detected, the image is rotated before OCR processing.

If orientation detection fails, the scraper uses a 180-degree rotation as a fallback.

This helps improve OCR accuracy when processing scanned project documents.

1.6 Identifying Individual Projects

After OCR has been performed, the resulting text is processed using regular expressions.

The parser looks for project numbers such as:

1.
2.
3.
4.
5.

These numbers are used to identify individual project records.

The parser also ignores numbers that are outside the expected project range and ignores header information that is not part of an actual project.

Each identified project is then processed independently.

1.7 Extracting CDF Project Information

Each project is mapped into a predefined structure containing the following fields:

Field	Description
Constituency	Mufulira constituency associated with the project
Project_No	Project number

Engineers_Estimate_ZMW	Engineer's estimated project cost
Approved_Amount_ZMW	Approved project amount
Justification	Project justification
Source_File	Original PDF filename

Financial values are identified using regular expressions.

1.13 CDF Output Dataset

The final CDF dataset is saved as:

db-unza26-csc4792-mufulira_cdf_projects.csv

The dataset uses a pipe (|) delimiter:

Constituency|Project_No|Ward|Project_Name|Sector|Scope_of_Work|Quantity|Location|Proposed_Date|Duration|Engineers_Estimate_ZMW|Approved_Amount_ZMW|Justification|Source_File

The pipe delimiter was used because project descriptions and other text fields can contain commas.

Therefore, using | makes the separation between columns explicit and avoids problems caused by commas occurring naturally within the data.

5. Difference Between CDF and Budget Extraction

The two data sources required different extraction techniques because the documents have different structures.

CDF Documents

The CDF documents are scanned documents.

Therefore, OCR was required:

PDF
 ↓
Image
 ↓
OCR
 ↓
Text
 ↓
Regex Parsing
 ↓
Structured Data
Budget Documents

The budget documents contain structured tables.

Therefore, direct table extraction was used:

PDF
 ↓
Table Extraction
 ↓
Raw Tables
 ↓
Data Cleaning
 ↓
Structured Data

This approach avoids unnecessarily applying OCR to documents where structured table extraction is available.

7. CSV Delimiter

All final datasets use the pipe character:

|

instead of a comma.

For example:

Constituency|Project_No|Ward|Project_Name|Sector|Scope_of_Work|Quantity|Location|Proposed_Date|Duration|Engineers_Estimate_ZMW|Approved_Amount_ZMW|Justification|Source_File

The pipe delimiter was selected because many fields contain commas naturally.

For example, a financial amount may appear as:

1,500,000.00

and a project description may also contain commas.

Using a pipe delimiter prevents these commas from being interpreted as column separators.
