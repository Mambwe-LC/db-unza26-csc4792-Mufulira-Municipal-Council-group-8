```python
import pandas as pd
import random

INPUT_FILE = "cleaned_datasets/cleaned_datasets/db-unza26-csc4792-mufulira_health_facilities.csv"

# Read the CSV
df = pd.read_csv(INPUT_FILE, sep="|")

# Check Constituency column
if "Constituency" not in df.columns:
    raise ValueError(
        f"Column 'Constituency' was not found. "
        f"Available columns: {df.columns.tolist()}"
    )

# The three constituencies
constituencies = [
    "Kankoyo",
    "Kantanshi",
    "Mufulira Central"
]

# Make results reproducible
random.seed(2026)

# Assign a random constituency to every row
df["Constituency"] = [
    random.choice(constituencies)
    for _ in range(len(df))
]

# Drop Catchment Population if it exists
if "Catchment Population" in df.columns:
    df.drop(columns=["Catchment Population"], inplace=True)
    print("Dropped column: Catchment Population")
else:
    print("Column 'Catchment Population' was not found, so nothing was dropped.")

# Save back to the SAME file
df.to_csv(
    INPUT_FILE,
    sep="|",
    index=False,
    encoding="utf-8"
)

# Display results
print(f"\nUpdated {len(df)} rows.")
print("\nConstituency distribution:")
print(df["Constituency"].value_counts())

print("\nFinal columns:")
print(df.columns.tolist())

print(f"\nSaved successfully to: {INPUT_FILE}")
```
