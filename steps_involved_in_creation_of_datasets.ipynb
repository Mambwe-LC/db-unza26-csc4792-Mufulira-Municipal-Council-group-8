# Mufulira Municipal Council Data Extraction

## Overview

Data was collected from the Mufulira Municipal Council website and converted into structured, pipe-delimited CSV datasets for data mining.

The extraction covered:

- Constituency Development Fund (CDF) Community Projects
- Budgets and Revenue Streams
- Skills Development Bursary Applicants
- Governance and Integrated Development Plan (IDP) Data

---

## 1. Constituency Development Fund (CDF)

### 1.1 Data Source

CDF project information was obtained from the official Mufulira Municipal Council CDF page:

<https://www.mufuliracouncil.gov.zm/?page_id=792>

The page was used to identify the available Community Projects PDF documents for:

- Kantanshi
- Kankoyo
- Mufulira Central

Rejected project documents were excluded.

### 1.2 Extraction Process

The CDF extraction followed these steps:

1. Access the Mufulira Municipal Council CDF webpage.
2. Discover the Community Projects PDF links using Requests and BeautifulSoup.
3. Download each PDF directly from the website into memory without saving the PDF locally.
4. Convert the scanned PDF pages into images using pdf2image.
5. Apply OCR using pytesseract and Tesseract.
6. Use regular expressions and predefined ward/sector lists to identify project information.
7. Combine all extracted projects into a Pandas DataFrame.
8. Export the final dataset as a pipe-delimited CSV file.

### 1.3 CDF Fields

The extracted project information includes:

- Constituency
- Project Number
- Ward
- Project Name
- Sector
- Scope of Work
- Quantity
- Location
- Proposed Date
- Duration
- Engineer's Estimate
- Approved Amount
- Justification
- Source File

### 1.4 CDF Output

The final dataset is:

`db-unza26-csc4792-mufulira_cdf_projects.csv`

The dataset uses `|` as the delimiter.

## 2. Budgets and Revenue Streams

### 2.1 Data Sources

Budget information was extracted from the official Mufulira Municipal Council budget documents for:

- 2023
- 2024
- 2025
- 2026

The four official PDF documents were processed directly from their URLs.

### 2.2 Extraction Process

The budget extraction followed these steps:

1. Download the four official budget PDFs using Requests.
2. Process the PDFs without requiring locally stored copies.
3. Extract tables from the PDFs using pdfplumber.
4. Preserve the extracted tables in a raw dataset.
5. Identify relevant budget and revenue records.
6. Classify the records into categories such as fees and charges, local taxes/rates, LGEF and expenditure.
7. Extract revenue sources, budget amounts and budget statuses.
8. Combine the records into a final Pandas DataFrame.
9. Export the datasets as pipe-delimited CSV files.

### 2.3 Budget and Revenue Fields

The final dataset contains:

- record_id
- year
- data_type
- category
- revenue_source
- budget_amount
- budget_status
- description
- source_title
- source_url
- source_type

### 2.4 Budget Output

Two datasets were produced:

- `db-unza26-csc4792_mufulira_budget_raw_tables_2023_2026.csv`
- `db-unza26-csc4792_mufulira_budget_revenue_2023_2026.csv`

The first contains the raw tables extracted from the four budget documents. The second contains the cleaned and structured budget and revenue information.

Both datasets use `|` as the delimiter.

## 3. Skills Development Bursary Applicants

### 3.1 Data Sources

Skills Development bursary applicant information was extracted from official Mufulira Municipal Council CDF approved applicants PDF documents for:

- Kantanshi
- Kankoyo
- Mufulira Central

The documents covered the 2025 and 2026 Skills Development bursary applicants, including approved applicants and replacement applicants where applicable.

### 3.2 Extraction Process

The Skills Development extraction followed these steps:

1. Download the official Skills Development PDF documents directly from their URLs using Requests.
2. Process the PDFs in memory without saving the source PDFs locally.
3. Detect whether each PDF contains extractable digital text.
4. For digital PDFs, extract tables using pdfplumber.
5. For scanned PDFs, convert the pages into images using pdf2image.
6. Apply OCR using pytesseract and Tesseract to extract applicant information from scanned pages.
7. Identify applicant rows and map the extracted values to the predefined Skills Development fields.
8. Separate guardian/parent names and mobile phone numbers into individual fields.
9. Combine the extracted applicants from all constituencies and years into a Pandas DataFrame.
10. Remove duplicate applicant records where applicable.
11. Export the final dataset as a pipe-delimited CSV file.

### 3.3 Skills Development Fields

The extracted applicant information includes:

- Constituency
- No
- Ward
- Zone
- Name of Pupil
- NRC
- Gender
- Date of Birth
- Type of Vulnerability
- Study Programme
- Study Type
- Course Duration
- Name of Skills Institution
- Total TEVETA Fees
- Approved
- Guardian/Parent Name
- Guardian/Parent Phone
- Source File

### 3.4 Skills Development Output

The final dataset is:

`db-unza26-csc4792-mufulira_cdf_skills_applicants.csv`

The dataset uses `|` as the delimiter.

## 4. Governance and Integrated Development Plan (IDP) Data

### 4.1 Data Source

Governance-related data was extracted from the official Mufulira Municipal Council Integrated Development Plan (IDP) document.

The IDP contains information covering the Mufulira District development planning period, including administrative structures, demographic information, public consultation issues, health facilities and planned capital investments.

### 4.2 Extraction Process

The Governance and IDP extraction followed these steps:

1. Download the official Mufulira Municipal Council IDP PDF directly from its URL using Requests.
2. Process the PDF in memory without saving the source PDF locally.
3. Extract structured tables from the PDF using pdfplumber.
4. Detect relevant tables using their headers and table structure.
5. Extract the raw cell values from the identified tables without cleaning or standardizing the extracted values.
6. Preserve the extracted records in Pandas DataFrames.
7. Export each dataset as a pipe-delimited CSV file.

No duplicate removal, value standardization, generated identifiers or manual records were applied during the extraction process.

### 4.3 Governance and IDP Datasets

Four datasets were extracted:

- Administrative Wards and Demographics
- Health Facilities
- Ward Public Consultation Issues
- Master Capital Investment Framework

### 4.4 Governance and IDP Fields

The Administrative Wards and Demographics dataset includes:

- Constituency
- Ward Name
- Number of Households
- Total Population
- Male Population
- Female Population
- Poverty Level Status

The Health Facilities dataset includes:

- Facility ID
- Facility Name
- Facility Type
- Ward
- Ownership
- Operational Status
- Catchment Population
- Constituency

The Ward Public Consultation Issues dataset includes:

- Ward Name
- Challenges
- Solutions
- Sector
- Constituency

The Master Capital Investment Framework dataset includes:

- Sector
- Main Objective
- Specific Objective or Program
- Project Description
- Cost ZMW
- Proposed Source of Funds
- Location
- Ward
- Responsible Agency
- Implementation Period

### 4.5 Governance and IDP Outputs

Four datasets were produced:

- `db-unza26-csc4792-mufulira_administrative_wards_demographics.csv`
- `db-unza26-csc4792-mufulira_health_facilities.csv`
- `db-unza26-csc4792-mufulira_ward_public_consultation_issues.csv`
- `db-unza26-csc4792-mufulira_master_capital_investment_framework.csv`

All four datasets use `|` as the delimiter.

## 6. Overall Process

### CDF

```
Website => PDF Links => PDF in Memory => OCR => Data Extraction => Pandas => CSV
```

### Budget

```
PDF URLs => PDF in Memory => Table Extraction => Data Cleaning => Pandas => CSV
```

### Skills Development

```
PDF URLs => PDF in Memory => Digital Table Extraction / OCR => Data Extraction => Pandas => CSV
```

### Governance and IDP

```
PDF URL => PDF in Memory => Table Extraction => Raw Data Extraction => Pandas => CSV
```

All final datasets are stored as pipe-delimited (`|`) CSV files for use in subsequent data mining and analysis.
